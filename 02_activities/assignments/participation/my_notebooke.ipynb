```python
import numpy as np 
import matplotlib.pyplot as plt 
import pandas as pd
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50) 
fig, ax = plt.subplots(figsize=(6,4), facecolor='#2b0b17')
ax.set_facecolor('#0b0b0b')

# Halloween-themed histogram
n, bins, patches = ax.hist(y, bins=10, color='#ff7518', edgecolor='black', linewidth=1.2, alpha=0.95)
for p in patches:
	p.set_hatch('///')

# styling
ax.set_title('Spooky Data Distribution', color='#ff7518', fontsize=16, weight='bold')
ax.tick_params(colors='#ffb366')
for spine in ax.spines.values():
	spine.set_color('#ff7518')

# pumpkin annotation (may render as emoji in supported environments)
ax.text(0.95, 0.95, '🎃', transform=ax.transAxes, ha='right', va='top', fontsize=24)

fig.tight_layout()
```

```python
# define font attributes for both axes of graph
font1 = {'family':'ransom note','color':'blue','size':20}
font2 = {'family':'comic sans ms','color':'darkred','size':15}
# create the visualization
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)
# paint/add description labels
ax.set_title('Total Growth over time', fontdict=font1, loc='left')
ax.set_xlabel('Years since start', fontdict=font2)
ax.set_ylabel('Total Growth', fontdict=font2)
# fig.tight_layout()
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y, marker="*", color='purple')
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y, marker="*", color='purple', linewidths=1, linestyle='--')

```

```python
# set size
fig, ax = plt.subplots(figsize=(5,3))
# define styles and paint the graph
ax.plot(x,y,
        marker = '*',
        markersize = 12,
        color = '#7425b9',
        linestyle = '--',
        linewidth = 2,
        markeredgecolor = '#fa9359',
        markerfacecolor = '#000000')
# modifying styles for the grid lines
ax.grid(color='green', linewidth=2, linestyle='--')
```
